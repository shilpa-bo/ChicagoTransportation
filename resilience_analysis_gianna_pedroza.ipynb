```python
import networkx as nx
import random
import matplotlib.pyplot as plt

def parse_tntp(file_path):
    """Parse a TNTP network file into a NetworkX graph."""
    G = nx.DiGraph()
    reading_data = False
    with open(file_path, 'r') as f:
        for line in f:
            line = line.strip()
            if line.startswith('<END OF METADATA>'):
                reading_data = True
                continue
            if not reading_data or line.startswith('~') or line.startswith('<') or not line:
                continue

            parts = line.split()
            init_node = int(parts[0])
            term_node = int(parts[1])
            capacity = float(parts[2])  # Column 3 is capacity
            length = float(parts[3])    # Column 4 is length (miles)
            free_flow_time = float(parts[4])  # Column 5 is free-flow travel time
            G.add_edge(init_node, term_node, capacity=capacity, length=length, weight=free_flow_time)

    return G

def compute_centrality_measures(G):
    """Compute centrality measures to identify critical nodes."""
    betweenness = nx.betweenness_centrality(G, weight='weight')
    eigenvector = nx.eigenvector_centrality_numpy(G, weight='weight')
    return betweenness, eigenvector

def simulate_random_failures(G, num_removals=10):
    """Simulate random node failures and measure impact on network connectivity."""
    G_copy = G.copy()
    largest_components = []

    for _ in range(num_removals):
        node_to_remove = random.choice(list(G_copy.nodes()))
        G_copy.remove_node(node_to_remove)
        largest_components.append(len(max(nx.strongly_connected_components(G_copy), key=len)))

    return largest_components

def simulate_targeted_attacks(G, num_removals=10):
    """Simulate targeted attacks by removing highest betweenness centrality nodes first."""
    G_copy = G.copy()
    betweenness, _ = compute_centrality_measures(G_copy)
    sorted_nodes = sorted(betweenness, key=betweenness.get, reverse=True)[:num_removals]

    largest_components = []
    for node in sorted_nodes:
        G_copy.remove_node(node)
        largest_components.append(len(max(nx.strongly_connected_components(G_copy), key=len)))

    return largest_components

def plot_resilience(random_failures, targeted_attacks):
    """Plot the network robustness under random failures vs. targeted attacks."""
    plt.figure(figsize=(8,5))
    plt.plot(range(len(random_failures)), random_failures, label='Random Failures', marker='o')
    plt.plot(range(len(targeted_attacks)), targeted_attacks, label='Targeted Attacks', marker='s')
    plt.xlabel("Number of Nodes Removed")
    plt.ylabel("Size of Largest Connected Component")
    plt.title("Node Failure Resilience Analysis")
    plt.legend()
    plt.show()
```

```python
# Example Usage
G = parse_tntp("/content/ChicagoSketch_net.tntp")  # Replace with actual file
rand_fail = simulate_random_failures(G, num_removals=20)
targ_att = simulate_targeted_attacks(G, num_removals=20)
plot_resilience(rand_fail, targ_att)
```

### Node Failure Resilience Analysis Plot
Compares the impact of random failures (blue line) and targeted attacks (orange line) on the size of the largest connected component in the Chicago Sketch transportation network.

#### Observations:
- Random Failures (Blue Line): As we randomly remove nodes, the network size decreases gradually. This suggests that the transportation network is somewhat resilient to random disruptions (e.g., accidents, small-scale road closures).
- Targeted Attacks (Orange Line): When highly central nodes (determined by betweenness centrality) are removed first, the network shrinks more slowly at the start. Then, after around 10 removals, the decline becomes steeper, so removing these key nodes significantly disrupts connectivity.

#### Comparison: Targeted Attacks vs. Random Failures
Random failures cause a steady decline, meaning that the network is well-distributed with no single point of failure. Targeted attacks degrade the network more efficientlythis confirms that certain roads or intersections are critical to overall connectivity. The difference between the curves shows that the network is not highly robust to strategic failures.

#### Implications for Urban Planning
- Identify & Strengthen Critical Nodes: The nodes removed in targeted attacks should be reinforced (e.g., alternative routes, redundancy).
- Disaster & Emergency Response: Authorities should focus on protecting high-betweenness nodes to minimize disruption.
- Network Design for Resilience: If a new metro line or road expansion is planned, it should help decentralize the most important nodes.

```python
def simulate_random_edge_failures(G, num_removals=10):
    """Simulate random edge failures and measure impact on network connectivity."""
    G_copy = G.copy()
    largest_components = []

    for _ in range(num_removals):
        edge_to_remove = random.choice(list(G_copy.edges()))
        G_copy.remove_edge(*edge_to_remove)
        if nx.is_strongly_connected(G_copy):
            largest_components.append(len(max(nx.strongly_connected_components(G_copy), key=len)))
        else:
            largest_components.append(0)

    return largest_components

def simulate_targeted_edge_attacks(G, num_removals=10):
    """Simulate targeted attacks by removing highest betweenness centrality edges first."""
    G_copy = G.copy()
    edge_betweenness = nx.edge_betweenness_centrality(G_copy, weight='weight')
    sorted_edges = sorted(edge_betweenness, key=edge_betweenness.get, reverse=True)[:num_removals]

    largest_components = []
    for edge in sorted_edges:
        G_copy.remove_edge(*edge)
        if nx.is_strongly_connected(G_copy):
            largest_components.append(len(max(nx.strongly_connected_components(G_copy), key=len)))
        else:
            largest_components.append(0)

    return largest_components

def plot_edge_resilience(random_failures, targeted_attacks, title="Network Resilience Analysis"):
    """Plot the network robustness under random failures vs. targeted attacks."""
    plt.figure(figsize=(8,5))
    plt.plot(range(len(random_failures)), random_failures, label='Random Failures', marker='o')
    plt.plot(range(len(targeted_attacks)), targeted_attacks, label='Targeted Attacks', marker='s')
    plt.xlabel("Number of Edges Removed")
    plt.ylabel("Size of Largest Connected Component")
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
rand_edge_fail = simulate_random_edge_failures(G, num_removals=20)
targ_edge_att = simulate_targeted_edge_attacks(G, num_removals=20)
plot_edge_resilience(rand_edge_fail, targ_edge_att, title="Edge Failure Resilience Analysis")
```

### Edge Failure Resilience Analysis Plot
Compares how the size of the largest connected component in the network decreases when edges (road segments) are removed randomly versus in a targeted manner.

#### Observations:
- Random Edge Failures (Blue Line): The network remains stable until around 78 edge removals.Then, the largest component collapses suddenly. This suggests that the network has redundancy but also some critical links that, when removed, cause a rapid breakdown.
- Targeted Edge Attacks (Orange Line): The network remains almost unaffected, even after 20 targeted edge removals. This suggests that even the highest betweenness edges are not single points of failure. The network is robust against edge-based attacks, unlike node-based failures.

#### Comparison: Nodes vs. Edges
In node failures, targeted attacks were more damaging, and in edge failures, the network seems more resistant to even targeted removals. This means the network heavily depends on key intersections (node-based vulnerability) rather than individual roads (edge-based strength). Even if high-capacity roads are removed, alternative routes keep the network functional, but if certain critical nodes fail, the whole system can collapse.
